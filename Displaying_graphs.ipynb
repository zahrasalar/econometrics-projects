```python
import numpy as np
```

```python
import math
```

```python
import matplotlib.pyplot as plt
```

```python
x = np.linspace(0,2*math.pi)
```

### numpy.linspace(start, stop, num=50, endpoint=True, retstep=False, dtype=None, axis=0, *, device=None)
### endpointbool, optional : If True, stop is the last sample. Otherwise, it is not included. Default is True.
###  retstepbool, optional : If True, return (samples, step), where step is the spacing between samples
###  dtype : integer; float

```python
print(x)
```

```python
plt.plot(x, np.sin(x), 'm^', label=r'$\sin(x)$')
plt.plot(x, np.cos(x), 'ro', label=r'$\cos(x)$')
plt.title(r'Two plots in a graph')
plt.legend()
```

در Matplotlib، هر چیزی که بین $ قرار بگیرد، به صورت LaTeX math mode تفسیر می‌شود.
