#Plotting a price and volume chart


```python
prices = gold['Close']
prices.head()
```

```python
prices = gold['Close']['GC=F']
volumes = gold['Volume']['GC=F']
prices.head()
```

```python
volumes.tail()
```

```python
print(type(prices))
print(type(volumes))
```

```python
# The top plot consisting of dailing closing prices
top = plt.subplot2grid((4, 4), (0, 0), rowspan=3, colspan=4)
top.plot(prices.index, prices, label='Close')
plt.title('Gold Close Price from 2025 - Present')
plt.xticks(rotation=45)
plt.legend(loc=2)

# The bottom plot consisting of daily trading volume
bottom = plt.subplot2grid((4, 4), (3,0), rowspan=1, colspan=4)
bottom.bar(volumes.index, volumes)
plt.title('Gold Daily Trading Volume')
plt.xticks(rotation=45)
plt.gcf().set_size_inches(12, 8)
plt.subplots_adjust(hspace=0.75)
```

# 'Inflation Comparison: Iran vs USA'

```python
!pip install pandas_datareader
```

```python
import pandas as pd
import pandas_datareader.data as web
import matplotlib.pyplot as plt
```

```python
# USA inflation
USA = web.DataReader('FPCPITOTLZGUSA', 'fred', start='2010-01-01')
# Iran inflation
Iran = web.DataReader('FPCPITOTLZGIRN','fred', start='2010-01-01')
#ترکیب دو DataFrame
inflation = pd.concat([USA, Iran], axis=1)

inflation.columns = ['USA Inflation','Iran Inflation']
inflation
```

```python
plt.figure(figsize=(12,6))

plt.plot(inflation.index, inflation['USA Inflation'],label='USA')

plt.plot(inflation.index, inflation['Iran Inflation'],label='Iran')


plt.title('Inflation Comparison: Iran vs USA')

plt.xlabel('Year')

plt.ylabel('Inflation Rate (%)')

plt.legend()

plt.grid()

plt.show()
```

## Plotting returns

```python
!pip install quandl
```

```python
prices = gold['Close']['GC=F']
daily_changes = prices.pct_change(periods=1)
daily_changes.plot()
```

##Plotting cummulative returns

```python
# cumsum() مخفف cumulative sum یعنی جمع تجمعی است.
df_cumsum = daily_changes.cumsum()
df_cumsum.plot();
```

## Plotting Histograms

```python
daily_changes.hist(bins=50, figsize=(8, 4));
```

```python
daily_changes.describe()
```

## Plotting volatility

```python
df_std = daily_changes.rolling(window=30, min_periods=30).std()
df_std.plot();
```
