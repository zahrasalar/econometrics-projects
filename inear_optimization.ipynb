## A simple linear optimization problem

```python
import pulp
```

```python
#x≥0
#یعنی در مدل PuLP متغیری به اسم x ایجاد می‌شود.
# lowBound=0    این یعنی حد پایین (minimum value) این متغیر صفر است.
x = pulp.LpVariable("x", lowBound=0)
y = pulp.LpVariable("y", lowBound=0)
```

```python
problem = pulp.LpProblem("A simple maximization objective", pulp.LpMaximize)
problem += 3*x + 2*y, "The objective function"
problem += 2*x + y <= 100, "1st constraint"
problem += x + y <= 80, "2st constraint"
problem += x <= 40, "3rd constraint"
problem.solve()
problem
```

```python
print("Maximization Results:")
for variable in problem.variables():
    print(variable.name, '=', variable.varValue)
```

## show plot

```python
import numpy as np
import matplotlib.pyplot as plt
x = np.linspace(0,50,500)
y1 = 100-2*x
y2 = 80-x
plt.plot(x,y1, label='2x + y = 100')
plt.plot(x,y2, label='x + y = 80')

plt.axvline(x=40, label='x = 40')
plt.scatter(20, 60, s=100, label='Optimal point (20,60)')

plt.text(20, 60, '  (20, 60)', fontsize=12)

plt.xlim(0, 100)
plt.ylim(0, 100)


plt.xlabel('x')
plt.ylabel('y')


plt.title('Linear Programming Problem')


plt.grid(True)


plt.legend()


plt.show()
```
